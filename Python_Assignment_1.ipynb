```python
input_str = "Hello Nasrin, Welcome to Python Programming"
print("First character: ",input_str[0])
print("Last character: ",input_str[-1])
print("First 5 characters: ",input_str[:5])
print("Last 11 characters: ",input_str[-12:])
print("The string in reverse: ",input_str[::-1])
print(input_str[25:31])
```

```python
strM = "Python beginner tutorial"
print("Upper case: ",strM.upper())
print("Lower case: ",strM.lower())
print("Original string: ",strM.capitalize())
print("The number of occurences of t: ",strM.count("t"))
print("String after applying replace: ",strM.replace("Python","Machine Learning"))
```

# Tuples

```python
tuple_1 = (10,20,30)
tuple_2 = (40,50,60)
t_combine = tuple_1 + tuple_2
print("The combined tuple: ",t_combine)

print("Repeat: ",t_combine*3)

print("The third element is: ",t_combine[2])

print("The first 3 elements are: ",t_combine[:3])

print("The last 3 elements are: ",t_combine[-3:])

```
